```python
// =====================================
// CAYLEY-HAMILTON VERIFIER
// FRONTEND JAVASCRIPT
// =====================================


// Create matrix inputs when page loads
window.onload = function () {

    createMatrix();

};


// =====================================
// CREATE MATRIX
// =====================================

function createMatrix() {

    const size =
        parseInt(
            document.getElementById("matrixSize").value
        );

    const container =
        document.getElementById("matrixInput");

    container.innerHTML = "";


    for (let i = 0; i < size; i++) {

        const row =
            document.createElement("div");

        row.className = "matrix-row";


        for (let j = 0; j < size; j++) {

            const input =
                document.createElement("input");

            input.type = "number";

            input.value = 0;

            input.id = `a${i}${j}`;

            row.appendChild(input);

        }

        container.appendChild(row);

    }

}


// =====================================
// READ MATRIX
// =====================================

function getMatrix() {

    const size =
        parseInt(
            document.getElementById("matrixSize").value
        );

    let matrix = [];


    for (let i = 0; i < size; i++) {

        let row = [];

        for (let j = 0; j < size; j++) {

            const value =
                parseFloat(
                    document.getElementById(
                        `a${i}${j}`
                    ).value
                );

            row.push(value);

        }

        matrix.push(row);

    }

    return matrix;

}


// =====================================
// DETERMINANT
// =====================================

function determinant(A) {

    const n = A.length;


    // 2 × 2
    if (n === 2) {

        return (
            A[0][0] * A[1][1]
            -
            A[0][1] * A[1][0]
        );

    }


    // 3 × 3
    if (n === 3) {

        return (

            A[0][0] *
            (
                A[1][1] * A[2][2]
                -
                A[1][2] * A[2][1]
            )

            -

            A[0][1] *
            (
                A[1][0] * A[2][2]
                -
                A[1][2] * A[2][0]
            )

            +

            A[0][2] *
            (
                A[1][0] * A[2][1]
                -
                A[1][1] * A[2][0]
            )

        );

    }

}


// =====================================
// CHARACTERISTIC POLYNOMIAL
// =====================================

function characteristicPolynomial(A) {

    const n = A.length;


    if (n === 2) {

        const trace =
            A[0][0] + A[1][1];

        const det =
            determinant(A);

        return [
            1,
            -trace,
            det
        ];

    }


    if (n === 3) {

        const trace =
            A[0][0]
            + A[1][1]
            + A[2][2];


        const c2 =

            A[0][0] * A[1][1]
            +
            A[0][0] * A[2][2]
            +
            A[1][1] * A[2][2]

            -

            A[0][1] * A[1][0]
            -
            A[0][2] * A[2][0]
            -
            A[1][2] * A[2][1];


        const det =
            determinant(A);


        return [
            1,
            -trace,
            c2,
            -det
        ];

    }

}


// =====================================
// POLYNOMIAL TO STRING
// =====================================

function polynomialToString(coefficients) {

    const n =
        coefficients.length - 1;

    let result = "";


    coefficients.forEach(
        (coefficient, index) => {

            const power = n - index;

            if (Math.abs(coefficient) < 1e-10)
                return;


            let sign =
                coefficient >= 0
                ? "+"
                : "-";


            let value =
                Math.abs(coefficient);


            if (result === "") {

                sign =
                    coefficient < 0
                    ? "-"
                    : "";

            }


            let term = "";


            if (power === 0) {

                term =
                    value.toFixed(2);

            }

            else {

                if (Math.abs(value - 1) > 1e-10) {

                    term =
                        value.toFixed(2);

                }


                term += "λ";


                if (power > 1) {

                    term += "^" + power;

                }

            }


            result +=
                " " + sign + " " + term;

        }
    );


    return result.trim();

}


// =====================================
// MATRIX MULTIPLICATION
// =====================================

function multiply(A, B) {

    const n = A.length;

    let result =
        Array.from(
            {length: n},
            () => Array(n).fill(0)
        );


    for (let i = 0; i < n; i++) {

        for (let j = 0; j < n; j++) {

            for (let k = 0; k < n; k++) {

                result[i][j] +=
                    A[i][k] * B[k][j];

            }

        }

    }

    return result;

}


// =====================================
// IDENTITY MATRIX
// =====================================

function identity(n) {

    let I =
        Array.from(
            {length: n},
            () => Array(n).fill(0)
        );


    for (let i = 0; i < n; i++) {

        I[i][i] = 1;

    }

    return I;

}


// =====================================
// MATRIX ADDITION
// =====================================

function add(A, B) {

    const n = A.length;

    return A.map(
        (row, i) =>
            row.map(
                (value, j) =>
                    value + B[i][j]
            )
    );

}


// =====================================
// SCALAR MULTIPLICATION
// =====================================

function scalarMultiply(A, scalar) {

    return A.map(
        row =>
            row.map(
                value =>
                    value * scalar
            )
    );

}


// =====================================
// MATRIX POWER
// =====================================

function matrixPower(A, power) {

    let result =
        identity(A.length);

    let base = A;


    while (power > 0) {

        if (power % 2 === 1) {

            result =
                multiply(
                    result,
                    base
                );

        }


        base =
            multiply(
                base,
                base
            );


        power =
            Math.floor(power / 2);

    }


    return result;

}


// =====================================
// FORMAT MATRIX
// =====================================

function formatMatrix(A) {

    return `
        <table style="
            border-collapse:collapse;
            margin:auto;
        ">

        ${A.map(row => `

            <tr>

                ${row.map(value => `

                    <td style="
                        padding:10px 18px;
                        border:1px solid #cbd5e1;
                    ">
                        ${Number(value).toFixed(4)}
                    </td>

                `).join("")}

            </tr>

        `).join("")}

        </table>
    `;

}


// =====================================
// MAIN CALCULATION
// =====================================

function calculateMatrix() {

    const A =
        getMatrix();

    const n =
        A.length;


    // Characteristic polynomial
    const coefficients =
        characteristicPolynomial(A);


    // Display characteristic equation
    document.getElementById(
        "characteristic"
    ).innerHTML =

        `<strong>
            p(λ) =
        </strong>

        ${polynomialToString(
            coefficients
        )}

        = 0`;


    // =================================
    // C-H VERIFICATION
    // =================================

    /*
       For a 2×2 matrix:

       p(A) =
       A² + c1 A + c2 I

       For a 3×3 matrix:

       p(A) =
       A³ + c1 A² + c2 A + c3 I
    */


    let pA =
        Array.from(
            {length:n},
            () => Array(n).fill(0)
        );


    for (
        let i = 0;
        i < coefficients.length;
        i++
    ) {

        const power =
            n - i;

        let term;


        if (power === 0) {

            term =
                scalarMultiply(
                    identity(n),
                    coefficients[i]
                );

        }

        else {

            term =
                scalarMultiply(
                    matrixPower(
                        A,
                        power
                    ),
                    coefficients[i]
                );

        }


        pA =
            add(
                pA,
                term
            );

    }


    const pAIsZero =
        pA.every(
            row =>
                row.every(
                    value =>
                        Math.abs(value) < 1e-9
                )
        );


    document.getElementById(
        "verification"
    ).innerHTML =

        pAIsZero

        ?

        `<span style="color:green;font-weight:bold;">
            ✓ VERIFIED: p(A) = 0
        </span>
        <br><br>
        ${formatMatrix(pA)}`

        :

        `<span style="color:red;font-weight:bold;">
            ✗ Verification failed
        </span>
        <br><br>
        ${formatMatrix(pA)}`;


    // =================================
    // INVERSE
    // =================================

    const det =
        determinant(A);


    if (Math.abs(det) < 1e-10) {

        document.getElementById(
            "inverse"
        ).innerHTML =

            `<span style="color:red;">
                A⁻¹ does not exist because det(A) = 0.
            </span>`;

    }

    else {

        document.getElementById(
            "inverse"
        ).innerHTML =

            `<strong>
                det(A) = ${det.toFixed(4)}
            </strong>

            <br><br>

            A⁻¹ exists.

            <br><br>

            <em>
                Full symbolic C-H inverse calculation
                is performed by the Python/SymPy backend.
            </em>`;

    }


    // =================================
    // A⁸
    // =================================

    const A8 =
        matrixPower(
            A,
            8
        );


    document.getElementById(
        "powerResult"
    ).innerHTML =

        formatMatrix(A8);


    // =================================
    // BENCHMARK
    // =================================

    const repetitions =
        5000;


    let start =
        performance.now();


    for (
        let i = 0;
        i < repetitions;
        i++
    ) {

        matrixPower(A, 8);

    }


    let end =
        performance.now();


    const directTime =
        (end - start) /
        repetitions;


    document.getElementById(
        "directTime"
    ).innerText =

        directTime.toFixed(6)
        + " ms";


    document.getElementById(
        "chTime"
    ).innerText =

        "Backend / Python";

}


// =====================================
// CLEAR
// =====================================

function clearMatrix() {

    createMatrix();


    document.getElementById(
        "characteristic"
    ).innerText =
        "Enter a matrix to calculate.";


    document.getElementById(
        "verification"
    ).innerText =
        "—";


    document.getElementById(
        "inverse"
    ).innerText =
        "—";


    document.getElementById(
        "powerResult"
    ).innerText =
        "—";


    document.getElementById(
        "chTime"
    ).innerText =
        "—";


    document.getElementById(
        "directTime"
    ).innerText =
        "—";

}
```
