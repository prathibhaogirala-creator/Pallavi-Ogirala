```python
# ============================================
# CELL 1: IMPORTS
# ============================================

import numpy as np
import sympy as sp
import pandas as pd
import matplotlib.pyplot as plt
import time

from IPython.display import display, Markdown

np.set_printoptions(precision=6, suppress=True)

print("✓ Cayley-Hamilton Verifier Tool (C7)")
print("✓ All libraries imported successfully.")
```

```python
# ============================================
# CELL 2: PROJECT INFORMATION
# ============================================

display(Markdown("""
# Cayley-Hamilton Verifier Tool (C7)

### Objective
Build an interactive mathematical tool for 2×2 and 3×3 matrices.

### The tool performs
1. Finds the characteristic polynomial.
2. Forms the characteristic equation.
3. Verifies the Cayley-Hamilton theorem: p(A) = 0.
4. Finds A⁻¹ using Cayley-Hamilton.
5. Finds A⁸ using Cayley-Hamilton.
6. Compares C-H computation with direct computation.
7. Displays mathematical verification and benchmark charts.

### Assumptions
- Only square 2×2 and 3×3 matrices are accepted.
- Characteristic polynomial:
  p(λ) = det(λI − A)
- A⁻¹ exists only when det(A) ≠ 0.
- Numerical equality is checked using a small tolerance.
"""))
```

```python
# ============================================
# CELL 3: USER INPUT
# ============================================

A_input = [
    [2, 1],
    [1, 2]
]

print("Input Matrix:")
print(np.array(A_input))
```

```python
# ============================================
# CELL 4: INPUT VALIDATION
# ============================================

A_np = np.array(A_input, dtype=float)

if A_np.ndim != 2:
    raise ValueError("Input must be a matrix.")

if A_np.shape[0] != A_np.shape[1]:
    raise ValueError("Matrix must be square.")

if A_np.shape[0] not in [2, 3]:
    raise ValueError("Only 2×2 and 3×3 matrices are supported.")

n = A_np.shape[0]

# Convert to SymPy
A = sp.Matrix(A_input)

print(f"✓ Valid {n}×{n} matrix")
print("\nA =")
display(A)
```

```python
# ============================================
# CELL 5: BASIC MATRIX INFORMATION
# ============================================

I = sp.eye(n)
det_A = sp.expand(A.det())

print("Identity Matrix I:")
display(I)

print("\ndet(A) =")
display(det_A)

if det_A == 0:
    print("\n⚠ A is singular.")
    print("A⁻¹ does not exist.")
else:
    print("\n✓ A is non-singular.")
    print("A⁻¹ exists.")
```

```python
# ============================================
# CELL 6: CHARACTERISTIC POLYNOMIAL
# ============================================

lam = sp.symbols('lambda')

char_poly = A.charpoly(lam)
p_lambda = sp.expand(char_poly.as_expr())
coefficients = char_poly.all_coeffs()

print("STEP 1: CHARACTERISTIC POLYNOMIAL")
print("----------------------------------")

print("Definition:")
print("p(λ) = det(λI − A)")

print("\nCharacteristic polynomial:")

display(sp.Eq(sp.Symbol("p"), p_lambda))

print("\nCoefficients:")
print(coefficients)
```

```python
# ============================================
# CELL 7: CHARACTERISTIC EQUATION
# ============================================

print("STEP 2: CHARACTERISTIC EQUATION")
print("--------------------------------")

display(sp.Eq(p_lambda, 0))

print("\nThis equation will now be used in the")
print("Cayley-Hamilton theorem.")
```

```python
# ============================================
# CELL 8: CAYLEY-HAMILTON EQUATION
# ============================================

print("STEP 3: CAYLEY-HAMILTON THEOREM")
print("--------------------------------")

print("The Cayley-Hamilton theorem states:")
print("Every square matrix satisfies its own characteristic equation.")

print("\nTherefore, if")

display(sp.Eq(p_lambda, 0))

print("\nthen replacing λ by A gives:")

# Construct symbolic p(A)
p_A = sp.zeros(n)

for i, coeff in enumerate(coefficients):
    power = n - i

    if power == 0:
        term = coeff * I
    else:
        term = coeff * (A ** power)

    p_A += term

display(sp.Eq(sp.Symbol("p(A)"), sp.expand(p_A)))
```

```python
# ============================================
# CELL 9: VERIFY p(A) = 0
# ============================================

print("STEP 4: VERIFY p(A) = 0")
print("------------------------")

p_A = sp.zeros(n)

for i, coeff in enumerate(coefficients):

    power = n - i

    if power == 0:
        term = coeff * I
        print(f"\nTerm: {coeff}I")
    else:
        term = coeff * (A ** power)
        print(f"\nTerm: {coeff}A^{power}")

    display(term)

    p_A += term

p_A = sp.simplify(p_A)

print("\nFinal result:")
display(p_A)

if p_A == sp.zeros(n):
    print("✓ VERIFIED: p(A) = 0")
else:
    print("✗ Verification failed.")
```

```python
# ============================================
# CELL 10: A^(-1) USING CAYLEY-HAMILTON
# ============================================

print("STEP 5: FINDING A⁻¹ USING CAYLEY-HAMILTON")
print("------------------------------------------")

if det_A == 0:

    print("det(A) = 0")
    print("Therefore A⁻¹ does not exist.")

    A_inverse_CH = None

else:

    c_n = coefficients[-1]

    # Construct numerator:
    # A^(n-1) + c1*A^(n-2) + ... + c(n-1)I

    numerator = sp.zeros(n)

    print("From p(A) = 0:")

    for i in range(n):

        coeff = coefficients[i]
        power = n - i

        term_power = power - 1

        if term_power == 0:
            term = coeff * I
            print(f"\n+ ({coeff})I")
        else:
            term = coeff * (A ** term_power)
            print(f"\n+ ({coeff})A^{term_power}")

        display(term)

        numerator += term

    print("\nTherefore:")
    print("A⁻¹ = −[numerator] / cₙ")

    A_inverse_CH = sp.simplify(-numerator / c_n)

    print("\nA⁻¹ using Cayley-Hamilton:")

    display(A_inverse_CH)

    print("\nDirect inverse:")

    A_inverse_direct = A.inv()

    display(A_inverse_direct)

    inverse_error = sp.simplify(
        A_inverse_CH - A_inverse_direct
    )

    print("\nDifference:")

    display(inverse_error)

    if inverse_error == sp.zeros(n):
        print("✓ INVERSE VERIFIED")
```

```python
# ============================================
# CELL 11: REDUCE λ⁸ USING CHARACTERISTIC EQUATION
# ============================================

print("STEP 6: REDUCING λ⁸ USING CAYLEY-HAMILTON")
print("------------------------------------------")

x = sp.symbols('x')

p_x = sp.Poly(
    p_lambda.subs(lam, x),
    x
)

print("Characteristic polynomial:")
display(p_x.as_expr())

# Polynomial division
quotient, remainder = sp.div(
    sp.Poly(x**8, x),
    p_x
)

print("\nPolynomial division:")
print("x⁸ = p(x) × q(x) + r(x)")

print("\nQuotient q(x):")
display(quotient.as_expr())

print("\nRemainder r(x):")
display(remainder.as_expr())

print("\nTherefore:")
print("x⁸ ≡ r(x)  (mod p(x))")
```

```python
# ============================================
# CELL 12: CALCULATE A⁸ USING C-H
# ============================================

def polynomial_to_matrix(poly_expr, matrix, variable):
    """
    Convert a polynomial in x into the
    corresponding polynomial in matrix A.
    """

    poly = sp.Poly(poly_expr, variable)

    result = sp.zeros(matrix.rows)

    for (power,), coeff in poly.terms():

        if power == 0:
            result += coeff * sp.eye(matrix.rows)
        else:
            result += coeff * (matrix ** power)

    return sp.simplify(result)


A8_CH = polynomial_to_matrix(
    remainder.as_expr(),
    A,
    x
)

print("Reduced polynomial:")
display(remainder.as_expr())

print("\nTherefore, replacing x by A:")

display(A8_CH)

print("\nA⁸ using Cayley-Hamilton:")
display(A8_CH)
```

```python
# ============================================
# CELL 13: DIRECT A⁸ VERIFICATION
# ============================================

print("STEP 7: VERIFY A⁸")
print("-----------------")

A8_direct = A ** 8

print("A⁸ calculated directly:")
display(A8_direct)

print("\nA⁸ calculated using Cayley-Hamilton:")
display(A8_CH)

A8_error = sp.simplify(A8_CH - A8_direct)

print("\nDifference:")
display(A8_error)

if A8_error == sp.zeros(n):
    print("✓ A⁸ VERIFIED")
else:
    print("✗ A⁸ verification failed.")
```

```python
# ============================================
# CELL 14: VERIFICATION ERRORS
# ============================================

def matrix_max_abs(M):
    """
    Maximum absolute element of a matrix.
    """
    values = [abs(float(v)) for v in M]
    return max(values) if values else 0.0


pA_error = matrix_max_abs(p_A)

if A_inverse_CH is not None:
    inverse_error_value = matrix_max_abs(
        A_inverse_CH - A.inv()
    )
else:
    inverse_error_value = None

A8_error_value = matrix_max_abs(
    A8_CH - A8_direct
)

print("VERIFICATION ERROR REPORT")
print("-------------------------")

print(f"p(A) error       : {pA_error}")

if inverse_error_value is not None:
    print(f"A⁻¹ error        : {inverse_error_value}")
else:
    print("A⁻¹ error        : Not applicable")

print(f"A⁸ error         : {A8_error_value}")

print("\nExpected result: errors should be 0 for exact SymPy calculations.")
```

```python
# ============================================
# CELL 15: FINAL RESULTS TABLE
# ============================================

results = {
    "Matrix Size": f"{n}×{n}",
    "det(A)": str(det_A),
    "Characteristic Polynomial": str(p_lambda),
    "p(A) = 0": "PASS" if p_A == sp.zeros(n) else "FAIL",
    "A⁻¹ Exists": "YES" if det_A != 0 else "NO",
    "A⁸ Verification": "PASS" if A8_error == sp.zeros(n) else "FAIL"
}

results_df = pd.DataFrame(
    results.items(),
    columns=["Property", "Result"]
)

display(Markdown("# Final Project Results"))
display(results_df)
```

```python
# ============================================
# CELL 16: BENCHMARK FUNCTIONS
# ============================================

def a8_using_cayley_hamilton(matrix_input):

    M = sp.Matrix(matrix_input)

    size = M.rows

    lam_local = sp.symbols('lambda')
    x_local = sp.symbols('x')

    char_poly_local = M.charpoly(lam_local)

    p_local = sp.Poly(
        char_poly_local.as_expr().subs(
            lam_local,
            x_local
        ),
        x_local
    )

    _, remainder_local = sp.div(
        sp.Poly(x_local**8, x_local),
        p_local
    )

    return polynomial_to_matrix(
        remainder_local.as_expr(),
        M,
        x_local
    )


def a8_direct_numpy(matrix_input):

    M = np.array(
        matrix_input,
        dtype=float
    )

    return np.linalg.matrix_power(M, 8)


def benchmark_method(method, matrix, repeats=1000):

    start = time.perf_counter()

    for _ in range(repeats):
        method(matrix)

    end = time.perf_counter()

    return (end - start) / repeats
```

```python
# ============================================
# CELL 17: BENCHMARK DATASET
# ============================================

benchmark_dataset = {

    "2×2 A": [
        [2, 1],
        [1, 2]
    ],

    "2×2 B": [
        [4, 1],
        [2, 3]
    ],

    "3×3 A": [
        [2, 1, 0],
        [1, 3, 1],
        [0, 1, 2]
    ],

    "3×3 B": [
        [3, 1, 2],
        [0, 2, 1],
        [1, 0, 4]
    ]
}

print("Benchmark matrices:")

for name, matrix in benchmark_dataset.items():

    print(f"\n{name}")
    print(np.array(matrix))
```

```python
# ============================================
# CELL 18: RUN BENCHMARK
# ============================================

REPEATS = 1000

benchmark_results = []

for name, matrix in benchmark_dataset.items():

    ch_time = benchmark_method(
        a8_using_cayley_hamilton,
        matrix,
        REPEATS
    )

    direct_time = benchmark_method(
        a8_direct_numpy,
        matrix,
        REPEATS
    )

    benchmark_results.append({
        "Matrix": name,
        "Size": f"{len(matrix)}×{len(matrix)}",
        "C-H Time (µs)": ch_time * 1e6,
        "Direct Time (µs)": direct_time * 1e6
    })


benchmark_df = pd.DataFrame(
    benchmark_results
)

print("BENCHMARK RESULTS")
display(
    benchmark_df.round(3)
)
```

```python
# ============================================
# CELL 19: CHARACTERISTIC COEFFICIENT CHART
# ============================================

coeff_values = [
    float(sp.N(c))
    for c in coefficients
]

labels = []

for i in range(len(coeff_values)):

    power = n - i

    if power == 0:
        labels.append("I")
    elif power == 1:
        labels.append("A")
    else:
        labels.append(f"A^{power}")


plt.figure(figsize=(8, 5))

plt.bar(
    labels,
    coeff_values
)

plt.axhline(
    0,
    linewidth=1
)

plt.title(
    "Characteristic Polynomial Coefficients"
)

plt.xlabel(
    "Matrix-power term"
)

plt.ylabel(
    "Coefficient"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()
```

```python
# ============================================
# CELL 20: BENCHMARK CHART
# ============================================

x_pos = np.arange(
    len(benchmark_df)
)

width = 0.35

plt.figure(figsize=(10, 5))

plt.bar(
    x_pos - width/2,
    benchmark_df["C-H Time (µs)"],
    width,
    label="Cayley-Hamilton"
)

plt.bar(
    x_pos + width/2,
    benchmark_df["Direct Time (µs)"],
    width,
    label="Direct NumPy"
)

plt.xticks(
    x_pos,
    benchmark_df["Matrix"]
)

plt.ylabel(
    "Average execution time (µs)"
)

plt.xlabel(
    "Test matrix"
)

plt.title(
    "Cayley-Hamilton vs Direct Matrix Power"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()
```

```python
# ============================================
# CELL 21: VERIFICATION ERROR CHART
# ============================================

error_labels = [
    "p(A)",
    "A⁸"
]

error_values = [
    pA_error,
    A8_error_value
]

plt.figure(figsize=(7, 5))

plt.bar(
    error_labels,
    error_values
)

plt.title(
    "Cayley-Hamilton Verification Errors"
)

plt.xlabel(
    "Verification"
)

plt.ylabel(
    "Maximum absolute error"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()

print("All errors should be zero.")
```

```python
// =====================================
// CAYLEY-HAMILTON VERIFIER
// FRONTEND JAVASCRIPT
// =====================================


// Create matrix inputs when page loads
window.onload = function () {

    createMatrix();

};


// =====================================
// CREATE MATRIX
// =====================================

function createMatrix() {

    const size =
        parseInt(
            document.getElementById("matrixSize").value
        );

    const container =
        document.getElementById("matrixInput");

    container.innerHTML = "";


    for (let i = 0; i < size; i++) {

        const row =
            document.createElement("div");

        row.className = "matrix-row";


        for (let j = 0; j < size; j++) {

            const input =
                document.createElement("input");

            input.type = "number";

            input.value = 0;

            input.id = `a${i}${j}`;

            row.appendChild(input);

        }

        container.appendChild(row);

    }

}


// =====================================
// READ MATRIX
// =====================================

function getMatrix() {

    const size =
        parseInt(
            document.getElementById("matrixSize").value
        );

    let matrix = [];


    for (let i = 0; i < size; i++) {

        let row = [];

        for (let j = 0; j < size; j++) {

            const value =
                parseFloat(
                    document.getElementById(
                        `a${i}${j}`
                    ).value
                );

            row.push(value);

        }

        matrix.push(row);

    }

    return matrix;

}


// =====================================
// DETERMINANT
// =====================================

function determinant(A) {

    const n = A.length;


    // 2 × 2
    if (n === 2) {

        return (
            A[0][0] * A[1][1]
            -
            A[0][1] * A[1][0]
        );

    }


    // 3 × 3
    if (n === 3) {

        return (

            A[0][0] *
            (
                A[1][1] * A[2][2]
                -
                A[1][2] * A[2][1]
            )

            -

            A[0][1] *
            (
                A[1][0] * A[2][2]
                -
                A[1][2] * A[2][0]
            )

            +

            A[0][2] *
            (
                A[1][0] * A[2][1]
                -
                A[1][1] * A[2][0]
            )

        );

    }

}


// =====================================
// CHARACTERISTIC POLYNOMIAL
// =====================================

function characteristicPolynomial(A) {

    const n = A.length;


    if (n === 2) {

        const trace =
            A[0][0] + A[1][1];

        const det =
            determinant(A);

        return [
            1,
            -trace,
            det
        ];

    }


    if (n === 3) {

        const trace =
            A[0][0]
            + A[1][1]
            + A[2][2];


        const c2 =

            A[0][0] * A[1][1]
            +
            A[0][0] * A[2][2]
            +
            A[1][1] * A[2][2]

            -

            A[0][1] * A[1][0]
            -
            A[0][2] * A[2][0]
            -
            A[1][2] * A[2][1];


        const det =
            determinant(A);


        return [
            1,
            -trace,
            c2,
            -det
        ];

    }

}


// =====================================
// POLYNOMIAL TO STRING
// =====================================

function polynomialToString(coefficients) {

    const n =
        coefficients.length - 1;

    let result = "";


    coefficients.forEach(
        (coefficient, index) => {

            const power = n - index;

            if (Math.abs(coefficient) < 1e-10)
                return;


            let sign =
                coefficient >= 0
                ? "+"
                : "-";


            let value =
                Math.abs(coefficient);


            if (result === "") {

                sign =
                    coefficient < 0
                    ? "-"
                    : "";

            }


            let term = "";


            if (power === 0) {

                term =
                    value.toFixed(2);

            }

            else {

                if (Math.abs(value - 1) > 1e-10) {

                    term =
                        value.toFixed(2);

                }


                term += "λ";


                if (power > 1) {

                    term += "^" + power;

                }

            }


            result +=
                " " + sign + " " + term;

        }
    );


    return result.trim();

}


// =====================================
// MATRIX MULTIPLICATION
// =====================================

function multiply(A, B) {

    const n = A.length;

    let result =
        Array.from(
            {length: n},
            () => Array(n).fill(0)
        );


    for (let i = 0; i < n; i++) {

        for (let j = 0; j < n; j++) {

            for (let k = 0; k < n; k++) {

                result[i][j] +=
                    A[i][k] * B[k][j];

            }

        }

    }

    return result;

}


// =====================================
// IDENTITY MATRIX
// =====================================

function identity(n) {

    let I =
        Array.from(
            {length: n},
            () => Array(n).fill(0)
        );


    for (let i = 0; i < n; i++) {

        I[i][i] = 1;

    }

    return I;

}


// =====================================
// MATRIX ADDITION
// =====================================

function add(A, B) {

    const n = A.length;

    return A.map(
        (row, i) =>
            row.map(
                (value, j) =>
                    value + B[i][j]
            )
    );

}


// =====================================
// SCALAR MULTIPLICATION
// =====================================

function scalarMultiply(A, scalar) {

    return A.map(
        row =>
            row.map(
                value =>
                    value * scalar
            )
    );

}


// =====================================
// MATRIX POWER
// =====================================

function matrixPower(A, power) {

    let result =
        identity(A.length);

    let base = A;


    while (power > 0) {

        if (power % 2 === 1) {

            result =
                multiply(
                    result,
                    base
                );

        }


        base =
            multiply(
                base,
                base
            );


        power =
            Math.floor(power / 2);

    }


    return result;

}


// =====================================
// FORMAT MATRIX
// =====================================

function formatMatrix(A) {

    return `
        <table style="
            border-collapse:collapse;
            margin:auto;
        ">

        ${A.map(row => `

            <tr>

                ${row.map(value => `

                    <td style="
                        padding:10px 18px;
                        border:1px solid #cbd5e1;
                    ">
                        ${Number(value).toFixed(4)}
                    </td>

                `).join("")}

            </tr>

        `).join("")}

        </table>
    `;

}


// =====================================
// MAIN CALCULATION
// =====================================

function calculateMatrix() {

    const A =
        getMatrix();

    const n =
        A.length;


    // Characteristic polynomial
    const coefficients =
        characteristicPolynomial(A);


    // Display characteristic equation
    document.getElementById(
        "characteristic"
    ).innerHTML =

        `<strong>
            p(λ) =
        </strong>

        ${polynomialToString(
            coefficients
        )}

        = 0`;


    // =================================
    // C-H VERIFICATION
    // =================================

    /*
       For a 2×2 matrix:

       p(A) =
       A² + c1 A + c2 I

       For a 3×3 matrix:

       p(A) =
       A³ + c1 A² + c2 A + c3 I
    */


    let pA =
        Array.from(
            {length:n},
            () => Array(n).fill(0)
        );


    for (
        let i = 0;
        i < coefficients.length;
        i++
    ) {

        const power =
            n - i;

        let term;


        if (power === 0) {

            term =
                scalarMultiply(
                    identity(n),
                    coefficients[i]
                );

        }

        else {

            term =
                scalarMultiply(
                    matrixPower(
                        A,
                        power
                    ),
                    coefficients[i]
                );

        }


        pA =
            add(
                pA,
                term
            );

    }


    const pAIsZero =
        pA.every(
            row =>
                row.every(
                    value =>
                        Math.abs(value) < 1e-9
                )
        );


    document.getElementById(
        "verification"
    ).innerHTML =

        pAIsZero

        ?

        `<span style="color:green;font-weight:bold;">
            ✓ VERIFIED: p(A) = 0
        </span>
        <br><br>
        ${formatMatrix(pA)}`

        :

        `<span style="color:red;font-weight:bold;">
            ✗ Verification failed
        </span>
        <br><br>
        ${formatMatrix(pA)}`;


    // =================================
    // INVERSE
    // =================================

    const det =
        determinant(A);


    if (Math.abs(det) < 1e-10) {

        document.getElementById(
            "inverse"
        ).innerHTML =

            `<span style="color:red;">
                A⁻¹ does not exist because det(A) = 0.
            </span>`;

    }

    else {

        document.getElementById(
            "inverse"
        ).innerHTML =

            `<strong>
                det(A) = ${det.toFixed(4)}
            </strong>

            <br><br>

            A⁻¹ exists.

            <br><br>

            <em>
                Full symbolic C-H inverse calculation
                is performed by the Python/SymPy backend.
            </em>`;

    }


    // =================================
    // A⁸
    // =================================

    const A8 =
        matrixPower(
            A,
            8
        );


    document.getElementById(
        "powerResult"
    ).innerHTML =

        formatMatrix(A8);


    // =================================
    // BENCHMARK
    // =================================

    const repetitions =
        5000;


    let start =
        performance.now();


    for (
        let i = 0;
        i < repetitions;
        i++
    ) {

        matrixPower(A, 8);

    }


    let end =
        performance.now();


    const directTime =
        (end - start) /
        repetitions;


    document.getElementById(
        "directTime"
    ).innerText =

        directTime.toFixed(6)
        + " ms";


    document.getElementById(
        "chTime"
    ).innerText =

        "Backend / Python";

}


// =====================================
// CLEAR
// =====================================

function clearMatrix() {

    createMatrix();


    document.getElementById(
        "characteristic"
    ).innerText =
        "Enter a matrix to calculate.";


    document.getElementById(
        "verification"
    ).innerText =
        "—";


    document.getElementById(
        "inverse"
    ).innerText =
        "—";


    document.getElementById(
        "powerResult"
    ).innerText =
        "—";


    document.getElementById(
        "chTime"
    ).innerText =
        "—";


    document.getElementById(
        "directTime"
    ).innerText =
        "—";

}
```
